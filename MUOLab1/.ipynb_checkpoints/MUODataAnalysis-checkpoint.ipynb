```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from pathlib import Path
```

## 1. Raw Data ##

```python
file_paths = ['MUO_data.txt', 'MUO_data_2.txt', 'MUO_data_3.txt', 'MUO_overnight.txt', 'MUO_overnight_10m.txt', 'MUO_overnight_2.txt', 'MUO_overnight_3.txt', 'MUO_overnight_4.txt',]

dataframes = []

for fp in file_paths: #goes through each file and concatenates them
    eachdata = pd.read_csv(fp, sep="\t", skiprows=3, engine="python")
    eachdata = eachdata.iloc[:, :6]
    dataframes.append(eachdata)
df = pd.concat(dataframes, ignore_index=True)

amp1 = df["Ch0_amp1 (V)"].to_numpy()
amp2 = df["Ch0_amp1 (V)"].to_numpy()
time1 = df["Ch0_time1 (s)"].to_numpy()
time2 = df["Ch0_time2 (s)"].to_numpy()
wid1 = df["Ch0_width1 (s)"].to_numpy()
wid2 = df["Ch0_width2 (s)"].to_numpy()
w1 = wid1*1e9 #into nanoseconds
w2 = wid2*1e9

df[:5] #sanity check
```

## 2. Distribution and Cuts ##

## 3. Digitizer Efficiency Calibration ##

```python
dfeff = pd.read_csv('MUO temp filtered pulse data digitizer efficiency.txt', sep="\t", skiprows=3, engine="python")

eff_time1 = dfeff["Ch0_time1 (s)"].to_numpy()
eff_time2 = dfeff["Ch0_time2 (s)"].to_numpy()

dteff = (eff_time2-eff_time1) *1e6

plt.hist(dteff, bins=50)
plt.xlabel("dt (µs)")
plt.ylabel("Counts per bin")
plt.title("Digitizer Efficiency - Time Difference Counts")
plt.show()
```

```python
tmin, tmax = 1.0, 40.0
counts_eff, new_bins = np.histogram(dteff, bins=bins_dt)
centers = 0.5*(new_bins[:-1]+new_bins[1:])

plt.step(centers, counts_eff)
plt.xlabel("dt (µs)")
plt.ylabel("Counts per bin")
plt.title("Digitizer Efficiency - Digitizer Bin Occupancy")
plt.show()
```

```python
#Correction:
bin_low, bin_high = 1.2, 38.0 #create a smaller window
lim = (centers>=bin_low)&(centers<=bin_high)
window = counts_eff[lim]

w_base = 10
corr_full = np.ones_like(counts_eff, dtype=float)

good = (window>=w_base)
if np.any(good):
    mean_ct = window[good].mean()
    corr = np.ones_like(window, dtype=float) #correction to baseline
    mask = (window>0)&good
    corr[mask] = mean_ct/window[mask]
    med = np.median(corr[mask]) if np.any(mask) else 1.0
    if med>0:
        corr /= med
    corr = np.clip(corr, 0.9, 1.1) #clipped to remove sharp deviations
    corr_full[lim] = corr
else:
    print("Warning. Not enough healthy bins.")

plt.figure(figsize=(7,4))
plt.scatter(centers[lim], corr_full[lim], s=20)
plt.axhline(1.0, color="gray", ls="--")
plt.xlabel("dt (µs)")
plt.ylabel("Correction factor")
plt.title("Digitizer Efficiency - Corrections")
plt.show()
```

```python
raw_counts, bin_edges = np.histogram(dt_after, bins=bins_dt)

eff_weights = np.ones_like(raw_counts, dtype=float)
eff_weights[lim] = corr_full[lim]

corrected_counts = raw_counts * eff_weights

plt.figure(figsize=(10,6))
plt.step(centers, raw_counts, color='red', label="Raw Cuts")
plt.step(centers, corrected_counts, color='blue', label="Digitizer Efficiency")
plt.xlabel("dt (µs)")
plt.ylabel("Counts per bin")
plt.title("Digitizer Efficiency - Efficiency Corrected Bin Occupancy")
plt.legend()
plt.show()
```

```python
#for finding systematic uncertainty
dt_fit = centers.astype(float)
D_corr = corrected_counts.astype(float)
D_uncorr = raw_counts.astype(float)

def fit_tau_from_counts(D_array, x_array):

    D_clip = np.clip(D_array, 0, None)
    
    if len(x_array) == len(D_array) + 1:
        x_array = 0.5 * (x_array[:-1] + x_array[1:])

    mask = (x_array >= 1.2) & (x_array <= 10.0) & (D_clip > 0)
    x_fit = x_array[mask]
    y_fit = np.log(D_clip[mask])

    slope, intercept = np.polyfit(x_fit, y_fit, 1)
    tau_fit = -1.0 / slope
    return tau_fit
```

## 4. Time Calibration from Digitizer Clock ##

```python
clock_files = {
    "1 us": "debug 1us cal 3.txt",
    "20 us": "debug 20us cal 3.txt",
    "30 us": "debug 30us cal 3.txt",
}

clock_time2 = "Ch0_time2 (s)"
clock_time1 = "Ch0_time1 (s)"

plt.figure(figsize=(8, 6))
colors = {"1 us": "tab:blue", "20 us": "tab:red", "30 us": "tab:green"}

median_clock=[]
std_clock=[]
nominal=[]
N=[]

for label, fname in clock_files.items():
    df = pd.read_csv(fname, sep="\t", header=3, engine="python")
    df = df.drop(columns=[c for c in df.columns if c.startswith("Ch1")])
    dtclock = (df[clock_time2].to_numpy()-df[clock_time1].to_numpy()) *1e6
    plt.hist(dtclock, bins=200, histtype="step", linewidth=1.5, label=label, color=colors[label])
    median_clock.append(float(np.median(dtclock))) #median to remove influence of outliers
    std_clock.append(float(np.std(dtclock)))
    nominal.append(float(label.split()[0]))
    N.append(len(dtclock))
    
plt.title(f"Time Calibration Overlay")
plt.xlabel("dt (µs)")
plt.ylabel("Counts per bin")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
y=np.array(median_clock, dtype=float)
stds=np.array(std_clock, dtype=float)
x=np.array(nominal, dtype=float)
N=np.array(N, dtype=float)

print(f"Medians (µs) → 1 µs: {y[0]:.3f} ± {stds[0]:.6f}, 20 µs: {y[1]:.3f} ± {stds[0]:.6f}, 30 µs: {y[2]:.3f} ± {stds[0]:.6f}.")

b, a = np.polyfit(x, y, 1)
y_fit = a + b*x
residuals = y-y_fit
#calib = (y-a)/b
#print(f"\nCalibration (actual = measured - a / b): calib_1us = {calib[0]:.6f} µs, calib_20us = {calib[1]:.6f} µs, calib_30us = {calib[2]:.6f} µs")
yerr = stds/(np.sqrt(N))

plt.plot(x, residuals, "bo")
plt.axhline(0, color="gray", linestyle="--", alpha=0.5)
plt.ylim(-1e-13, 1e-13)
plt.title(f"Residuals of Time Calibration")
plt.xlabel("Nominal (µs)")
plt.ylabel("Measured (µs)")

plt.tight_layout()
plt.show()
```

## 5. Semi-log Plot D with Background ##

```python
dt = centers_clock.astype(float)
D = corrected_counts_clock.astype(float)

#finding window
lower, upper = 12.0, 40.0
mask = (dt >= lower)&(dt <= upper)&(D>0)
assert mask.sum()>2

bg_mean = float(np.mean(D[mask]))
bg_std = float(np.std(D[mask], ddof=1))
bg_sem = float(bg_std/np.sqrt(mask.sum()))

x=dt[mask]
y=np.log(D[mask])
slope, intercept = np.polyfit(x, y, 1) #fitting the line
bg_D = np.exp(intercept+slope*dt) #mapping to linear space

D2 = D-bg_D
D2_lin = np.clip(D2, 0, None)
D2_log = np.where(D2>0, D2, np.nan)
#print(f"ln D = {intercept:.3f} + ({slope:.3f})*t  →  D_bg(t) = exp({intercept:.3f} + ({slope:.3f})*t)")

#semi-log with background linear fitted
plt.scatter(dt, D, s=5, color='green')
plt.plot(dt, bg_D, "--", label=f"Fitted Background: ln D = {intercept:.3f} + ({slope:.3f})*dt")
plt.yscale("log")
plt.xlabel("dt (µs)")
plt.ylabel("Counts per bin (log scale)")
plt.title("Semi-log Plot - Fitted Background")
plt.legend()
plt.show()

plt.scatter(dt, D2_lin, s=5)
plt.yscale("log")
plt.xlabel("dt (µs)")
plt.ylabel("Counts per bin (log scale)")
plt.title("Semi-log Plot - Background Removed")
plt.show()
```

```python
dt = centers_clock.astype(float)
D = corrected_counts_clock.astype(float)

mask_all = (dt >= 12) & (dt <= 40.0) & (D >0)

bg_mean = float(np.mean(D[mask_all]))
bg_std = float(np.std(D[mask_all], ddof=1))
bg_sem = float(bg_std/np.sqrt(mask_all.sum()))

def fit_tau_const_bg(B_val):
    
    D_sub = D-B_val
    D_sub = np.clip(D_sub, 0, None)
    
    #finding window
    lower, upper = 1.2, 10.0
    mask = (dt >= lower)&(dt <= upper)&(D_sub>0)
    assert mask.sum()>2

    x=dt[mask]
    y=np.log(D_sub[mask])
    slope, intercept = np.polyfit(x, y, 1) #fitting the line
    tau_fit = -1.0/slope
    return tau_fit
```

## 6. Semi-log Plot D<sub>2</sub> with Least-Squares Fit ##

```python
D2 = D2.astype(float)

#finding window
lower, upper = 1.2, 10.0
mask = (dt >= lower)&(dt <= upper)&(D2>0)
assert x.size >= 2

x=dt[mask]
y=np.log(D2[mask])
(slope, intercept), cov = np.polyfit(x, y, 1, cov=True) #fitting the line
slope_err = float(np.sqrt(cov[0,0]))
intercept_err = float(np.sqrt(cov[1,1]))

#lifetime
tau = -1.0/slope
tau_err = slope_err/(slope*slope)

print(f"Muon lifetime τ = {tau:.3f} ± {tau_err:.3f} µs")

xline=np.linspace(lower, upper, 200)
fit_line = np.exp(intercept+slope*xline)

#semi-log scale
plt.scatter(dt, D2_lin, s=5)
plt.plot(xline, fit_line, "-", color="orange", label=f"LST: ln(D2) = \n{intercept:.3f} ± {intercept_err:.3f} + ({slope:.3f} ± {slope_err:.3f}) * dt")
plt.yscale("log")
plt.xlim(0.5, 10.5)
plt.ylim(1*10^0, 10000*10^5)
plt.axvspan(lower, upper, color="k", alpha=0.06, label="Fit window")
plt.xlabel("dt (µs)")
plt.ylabel("Counts per bin (log scale)")
plt.title("LST of Muon Lifetime - Semi-log Plot")
plt.legend()
plt.show()

#linear scale
plt.step(dt, np.clip(D2, 0, None))
plt.plot(xline, fit_line, "-", color="orange", label=f"LST: ln(D2) = \n{intercept:.3f} ± {intercept_err:.3f} + ({slope:.3f} ± {slope_err:.3f}) * dt")
plt.xlim(0.5, 10.5)
plt.ylim(0, 30000)
plt.axvspan(lower, upper, color="k", alpha=0.06, label="Fit window")
plt.xlabel("dt (µs)")
plt.ylabel("Counts per bin")
plt.title("LST of Muon Lifetime - Linear Plot")
plt.legend()
plt.show()
```

```python
lower, upper = 2.0, 8.5
mask = (dt >= lower)&(dt <= upper)&(D2>0)

#normalize to same number of counts
def norm_amp_unweight(tau_val): 
    model = np.exp(-dt[mask]/tau_val)
    A = (D2[mask].sum())/(model.sum())
    return A
    
model_best = norm_amp_unweight(tau) * np.exp(-dt/tau)
model_upper = norm_amp_unweight(tau+tau_err) * np.exp(-dt/(tau+tau_err))
model_lower = norm_amp_unweight(max(tau-tau_err, 1e-9)) * np.exp(-dt/(max(tau-tau_err, 1e-9)))

#normalize with weighted least squares
win = (dt>=lower)&(dt<=upper)
dt_fit = dt[win]
D2_fit = D[win]
bg_fit = bg_D[win]
E_fit = np.exp(-dt_fit/tau)

weights = 1.0/np.clip(D2_fit, 1.0, None)
A_weighted = np.sum(weights*(D2_fit-bg_fit)*E_fit)/np.sum(weights*E_fit*E_fit)
model_fit = A_weighted*E_fit+bg_fit

#chi-squared
#sigma = 1.0 + np.sqrt(np.clip(D2_fit, 0.0, None)+0.75) #gehrels
sigma = np.sqrt(D2_fit+bg_fit)
res = (D2_fit-model_fit)/sigma
chi2 = float(np.sum(res**2))
dof = int(win.sum()-2)
red_chi2 = chi2 /max(dof, 1)

xline=np.linspace(lower, upper, 200)
fit_line = np.exp(intercept+slope*xline)

#semi-log scale
plt.figure(figsize=(8,7))
plt.scatter(dt, D2_lin, s=5)
plt.errorbar(dt, D2_lin, np.sqrt(D2_lin+bg_D), linestyle="")
plt.plot(xline, fit_line, "-", color="orange", label=f"Best LST Fit")
plt.plot(dt, model_lower,   "--", color="purple", label=f"τ − σ = {tau - tau_err:.3f} µs")
plt.plot(dt, model_upper,   "--", color="indigo", label=f"τ + σ = {tau + tau_err:.3f} µs")
plt.yscale("log")
plt.xlim(0.5, 10.5)
plt.ylim(10**1.5, 10**4.5)
plt.axvspan(lower, upper, color="k", alpha=0.06, label="Fit window")
plt.xlabel("dt (µs)")
plt.ylabel("Counts per bin (log scale)")
plt.title("LST of Muon Lifetime - Semi-log Plot")
plt.legend()
plt.show()
```

```python
def fit_tau_with_window(tmin, tmax):
    # subtract constant background from corrected spectrum
    D_sub = corrected_counts - bg_mean
    D_sub = np.clip(D_sub, 0, None)

    mask = (centers >= tmin) & (centers <= tmax) & (D_sub > 0)
    if np.sum(mask) < 3:
        return np.nan

    x = centers[mask]
    y = np.log(D_sub[mask])
    slope, intercept = np.polyfit(x, y, 1)
    tau_val = -1.0 / slope
    return tau_val

# various windows:
tau_main = fit_tau_with_window(1.2, 10.0)
tau_alt1 = fit_tau_with_window(1.5, 9.0)
tau_alt2 = fit_tau_with_window(2.0, 8.5)

print(f"Possible lifetimes from different windows: {tau_main:.6g}, {tau_alt1:.6g}, {tau_alt2:.6g}")
```

```python
def tau_from_hist(dt_array, bin_edges):
    # histogram with new bins
    counts, edges = np.histogram(dt_array, bins=bin_edges)
    centers_new = 0.5 * (edges[:-1] + edges[1:])

    D_sub = counts - bg_mean
    D_sub = np.clip(D_sub, 0, None)

    mask = (centers_new >= 1.2) & (centers_new <= 10.0) & (D_sub > 0)
    if np.sum(mask) < 3:
        return np.nan

    x = centers_new[mask]
    y = np.log(D_sub[mask])
    slope, intercept = np.polyfit(x, y, 1)
    tau_val = -1.0 / slope
    return tau_val

orig_min = bins_dt[0]
orig_max = bins_dt[-1]

# Method A: same number of bins but shifted by half a bin
bin_width = bins_dt[1] - bins_dt[0]
shifted_bins = np.arange(orig_min + 0.5*bin_width, orig_max + 0.5*bin_width, bin_width)

# Method B: coarser bins (2x width)
coarse_bins = np.arange(orig_min, orig_max + 2*bin_width, 2*bin_width)

tau_main_binning   = tau_from_hist(dt_after, bins_dt)         # your normal
tau_shift_binning  = tau_from_hist(dt_after, shifted_bins)
tau_coarse_binning = tau_from_hist(dt_after, coarse_bins)
```

## 7. Nuclear Capture ##

```python
free_decay = 1/2.197 #2.197 from pdg
exp_cap_decay = 1/26 #according to lab manual
eff_decay = 1/tau

cap_decay = 2*(eff_decay-free_decay) #1/2free_decay + 1/2(free_decay+cap_decay)=free_decay + 1/2cap_decay

err_eff_decay = tau_err/(tau**2)
err_cap_decay = err_eff_decay*2

uminus_lt = 1/(free_decay+cap_decay)

print(f"µ- capture rate = {cap_decay:.3g} ± {err_cap_decay:.3g} µs which is within one standard deviation of the estimated {exp_cap_decay:.3g}.")
print(f"The lifetime of µ- is {uminus_lt:.3g} which is less than the effective muon lifetime we found {tau:.3g} which is expected since the µ- decay faster.")
```

## 8. Uncertainty ##

```python
print(f"Muon lifetime τ = {tau:.3f} ± {tau_err:.3f} µs  χ² = {chi2:.1f},  χ²/dof = {red_chi2:.2f}")
print("The chi-square is very high and this is likely due to systematic errors as adjusting the window, using poisson statstics, or using different bins did not help reduce the chi-square value significantly.")
```

```python
#calib error: we assume 1% timing-scale uncertainty since there were only 3 points and calculating it gave an error of 3.9e-16 us
calib_err = tau * 0.01

#digit error: fn to find tau before and after being corrected is under question 3
tau_corr   = fit_tau_from_counts(D_corr, centers)
tau_uncorr = fit_tau_from_counts(D_uncorr, centers)
digit_err = abs(tau_corr - tau_uncorr)

#bg error: fn to find tau within one std under question 5
tau_plus = fit_tau_const_bg(bg_mean + bg_sem)
tau_minus = fit_tau_const_bg(bg_mean - bg_sem)
bg_err = 0.5 * abs(tau_plus - tau_minus)

#fit error: this fluctuates a lot based on the window we choose for the uncertainty calculation
taus = [t for t in [tau_main, tau_alt1, tau_alt2] if not np.isnan(t)]
fit_err = 0.5 * (max(taus) - min(taus)) if len(taus) >= 2 else 0.0

# binning error
taus_binning = [t for t in [tau_main_binning, tau_shift_binning, tau_coarse_binning] if not np.isnan(t)]
bin_err = 0.5 * (max(taus_binning) - min(taus_binning)) if len(taus_binning) >= 2 else 0.0

sys_sources = {
    "Time calibration scale": calib_err,
    "Digitizer efficiency correction": digit_err,
    "Background model": bg_err,
    "Fitting method (WLS vs Poisson)": fit_err,
    "Binning choice": bin_err,
}

tau_sys = np.sqrt(sum(v**2 for v in sys_sources.values())) #combining in quadrature

print("Systematic Uncertainty Budget (fractional, relative to τ):\n")
for k, v in sys_sources.items():
    print(f"  {k:30s} ±{v:.3g}")

print(f"\nSystematic uncertainty of τ = ±{tau_sys:.3f} µs")

print(f"\nFinal Muon Lifetime:")
print(f"  τ = {tau:.3f} ± {tau_err:.3f} ± {tau_sys:.3f} µs")
```

## 9. Fermi Coupling ##

```python
hbar = 6.582e-25 #GeV
gamma = hbar/(tau*(10**-6)) #decay rate in GeV
pi = 3.142 
m_u = 0.105658

G = np.sqrt((gamma * 192 * pi**3)/(m_u**5)) #measures strength of weak force assuming four-fermion contact interaction
G_err = 0.5*(tau_err/tau)*G #statistical error with G_F^2 proportional to -1/tau

print(f"{G:.4g} ± {G_err:.3g} which is more than the published value of 1.166e-05. Thus our measure lifetime is slightly too short due to our systematic errors.")
```

```python
G_sys = (0.5*(tau_sys/tau))*G

print(f"\nFinal Fermi Coupling Constant: G = {G:.3g} ± {G_err:.3g} ± {G_sys:.3g} GeV-2")
```
